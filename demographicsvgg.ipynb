```python
from typing import List
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from deepface import DeepFace
from deepface.models.Detector import FacialAreaRegion
from mtcnn import MTCNN

# ------------------------------
# Face Detector Wrapper
# ------------------------------
class MtCnnClient:
    def __init__(self):
        self.model = MTCNN()

    def detect_faces(self, img: np.ndarray) -> List[FacialAreaRegion]:
        results = []
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        detections = self.model.detect_faces(img_rgb)

        for det in detections:
            x, y, w, h = det["box"]
            confidence = det["confidence"]
            left_eye = det["keypoints"]["right_eye"]
            right_eye = det["keypoints"]["left_eye"]

            region = FacialAreaRegion(
                x=x, y=y, w=w, h=h,
                left_eye=left_eye, right_eye=right_eye,
                confidence=confidence
            )
            results.append(region)
        return results

# ------------------------------
# Preprocessing Face Crop
# ------------------------------
def preprocess_face(face_img):
    gray = cv2.cvtColor(face_img, cv2.COLOR_BGR2GRAY)
    equalized = cv2.equalizeHist(gray)
    processed = cv2.cvtColor(equalized, cv2.COLOR_GRAY2BGR)
    return processed

# ------------------------------
# Analyze a face with DeepFace
# ------------------------------
def analyze_face(img, face_region, scale=1.2, padding=40):
    x, y, w, h = face_region.x, face_region.y, face_region.w, face_region.h
    cx, cy = x + w//2, y + h//2

    # Expand box with scale & padding
    new_w, new_h = int(w*scale), int(h*scale)
    new_x = max(cx - new_w//2 - padding, 0)
    new_y = max(cy - new_h//2 - padding, 0)
    new_w = min(new_w + 2*padding, img.shape[1] - new_x)
    new_h = min(new_h + 2*padding, img.shape[0] - new_y)

    face_img = img[new_y:new_y+new_h, new_x:new_x+new_w]
    face_resized = cv2.resize(face_img, (224, 224))
    preprocessed_face = preprocess_face(face_resized)

    # Run DeepFace analysis
    results = DeepFace.analyze(
        img_path=preprocessed_face,
        actions=['age', 'gender', 'race', 'emotion'],
        detector_backend='skip',
        enforce_detection=False
    )
    if isinstance(results, list):
        results = results[0]
    return results, face_resized

# ------------------------------
# Annotate Image
# ------------------------------
def draw_annotations(img, detections, results_list):
    annotated_img = img.copy()
    for face_region, result in zip(detections, results_list):
        x, y, w, h = face_region.x, face_region.y, face_region.w, face_region.h
        label = f"{result['age']}y, {result['dominant_gender']}, {result['dominant_emotion']}"
        cv2.rectangle(annotated_img, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(annotated_img, label, (x, y-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
    return annotated_img

# ------------------------------
# Plot Dashboard
# ------------------------------
def plot_dashboard(results_list):
    num_faces = len(results_list)
    fig, axes = plt.subplots(num_faces, 4, figsize=(20, 5*num_faces))
    if num_faces == 1:
        axes = np.expand_dims(axes, axis=0)

    for i, result in enumerate(results_list):
        # Age
        axes[i,0].text(0.5, 0.5, f"{result.get('age', 'N/A')} years", fontsize=16, ha='center')
        axes[i,0].axis('off')
        axes[i,0].set_title(f"Face {i+1} - Age", fontsize=14)

        # Gender
        gender_probs = result.get('gender', {})
        axes[i,1].barh(list(gender_probs.keys()), [v*100 for v in gender_probs.values()], color='lightcoral')
        axes[i,1].set_xlim(0, 100)
        axes[i,1].set_xlabel("Probability (%)")
        axes[i,1].set_title(f"Face {i+1} - Gender", fontsize=14)

        # Emotion
        emotion_probs = result.get('emotion', {})
        axes[i,2].barh(list(emotion_probs.keys()), [v*100 for v in emotion_probs.values()], color='skyblue')
        axes[i,2].set_xlim(0, 100)
        axes[i,2].set_xlabel("Probability (%)")
        axes[i,2].set_title(f"Face {i+1} - Emotion", fontsize=14)

        # Ethnicity
        race_probs = result.get('race', {})
        axes[i,3].barh(list(race_probs.keys()), [v*100 for v in race_probs.values()], color='lightgreen')
        axes[i,3].set_xlim(0, 100)
        axes[i,3].set_xlabel("Probability (%)")
        axes[i,3].set_title(f"Face {i+1} - Ethnicity", fontsize=14)

    plt.tight_layout()
    plt.show()

# ------------------------------
# Main Pipeline
# ------------------------------
def main():
    print("Upload an image:")
    uploaded = files.upload()
    if not uploaded:
        print("No image uploaded.")
        return

    img_path = list(uploaded.keys())[0]
    img = cv2.imread(img_path)
    if img is None:
        raise ValueError("Error reading uploaded image.")

    detector = MtCnnClient()
    detections = detector.detect_faces(img)

    if not detections:
        print("No faces detected.")
        return

    print(f"Detected {len(detections)} face(s). Running analysis...\n")
    results_list = []

    for i, face in enumerate(detections, start=1):
        result, face_img = analyze_face(img, face)
        results_list.append(result)

        # Show individual face
        plt.figure(figsize=(3,3))
        plt.imshow(cv2.cvtColor(face_img, cv2.COLOR_BGR2RGB))
        plt.axis("off")
        plt.title(f"Face {i}")
        plt.show()

        # Print textual info
        print(f"--- Face {i} ---")
        print(f"Age: {result.get('age', 'N/A')}")
        print(f"Gender: {result.get('dominant_gender', 'N/A')} ({max(result['gender'].values()):.2f}%)")
        print(f"Emotion: {result.get('dominant_emotion', 'N/A')}")
        print("Race Probabilities:")
        for race, val in result.get('race', {}).items():
            print(f"  {race:15}: {val:.2f}%")
        print(f"Most Likely Race: {result.get('dominant_race', 'N/A')}")
        print("-"*40)

    # Annotated image
    annotated_img = draw_annotations(img, detections, results_list)
    plt.figure(figsize=(10,8))
    plt.imshow(cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.title("Annotated Image")
    plt.show()

    # Dashboard
    plot_dashboard(results_list)

# ------------------------------
if __name__ == "__main__":
    main()
```

```python
from typing import List
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from deepface import DeepFace
from deepface.models.Detector import FacialAreaRegion
from mtcnn import MTCNN

class MtCnnClient:
    def __init__(self):
        self.model = MTCNN()

    def detect_faces(self, img: np.ndarray) -> List[FacialAreaRegion]:
        results = []
        img_rgb = img[:, :, ::-1]  # Convert BGR -> RGB
        detections = self.model.detect_faces(img_rgb)

        if detections:
            for det in detections:
                x, y, w, h = det["box"]
                padding = 0.2
                x = max(0, int(x - w * padding / 2))
                y = max(0, int(y - h * padding / 2))
                w = int(w * (1 + padding))
                h = int(h * (1 + padding))
                confidence = det["confidence"]
                left_eye = det["keypoints"]["right_eye"]
                right_eye = det["keypoints"]["left_eye"]

                region = FacialAreaRegion(
                    x=x, y=y, w=w, h=h,
                    left_eye=left_eye, right_eye=right_eye,
                    confidence=confidence
                )
                results.append(region)
        return results

def preprocess_face(face_img):
    gray = cv2.cvtColor(face_img, cv2.COLOR_BGR2GRAY)
    equalized = cv2.equalizeHist(gray)
    processed = cv2.cvtColor(equalized, cv2.COLOR_GRAY2BGR)
    return processed

def analyze_demographics(img, face_region):
    x, y, w, h = face_region.x, face_region.y, face_region.w, face_region.h
    face_img = img[y:y+h, x:x+w]

    # Resize face to 224x224
    face_img_resized = cv2.resize(face_img, (224, 224))
    preprocessed_face = preprocess_face(face_img_resized)

    results = DeepFace.analyze(
        img_path=preprocessed_face,
        actions=['age', 'gender', 'race', 'emotion'],
        detector_backend='skip',
        enforce_detection=False
    )

    if isinstance(results, list):
        results = results[0]

    gender_confidence = max(results['gender'].values()) if "gender" in results else 0
    if gender_confidence < 70:
        results = DeepFace.analyze(
            img_path=face_img,
            actions=['age', 'gender', 'race', 'emotion'],
            detector_backend='retinaface',
            enforce_detection=False
        )
        if isinstance(results, list):
            results = results[0]

    return results, face_img_resized

def draw_annotations(img, detections, results_list):
    annotated_img = img.copy()

    for face_region, result in zip(detections, results_list):
        x, y, w, h = face_region.x, face_region.y, face_region.w, face_region.h
        label = f"{result['age']}y, {result['dominant_gender']}, {result['dominant_emotion']}"

        cv2.rectangle(annotated_img, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(
            annotated_img, label, (x, y-10),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2
        )

    return annotated_img

def main():
    print("Upload an image with one or more people:")
    uploaded = files.upload()

    if not uploaded:
        print("No image uploaded. Exiting.")
        return

    img_path = list(uploaded.keys())[0]
    img = cv2.imread(img_path)
    if img is None:
        raise ValueError("Error reading uploaded image.")

    detector = MtCnnClient()
    detections = detector.detect_faces(img)

    if len(detections) == 0:
        print("No faces detected in the image.")
        return

    print(f"Detected {len(detections)} face(s). Running demographics...\n")

    results_list = []
    for i, face in enumerate(detections, start=1):
        result, face_img = analyze_demographics(img, face)
        results_list.append(result)

        plt.figure(figsize=(3, 3))
        plt.imshow(cv2.cvtColor(face_img, cv2.COLOR_BGR2RGB))
        plt.axis("off")
        plt.title(f"Face {i}")
        plt.show()

        print(f"--- Face {i} Results ---")
        print(f"Age: {result.get('age', 'N/A')}")
        print(f"Gender: {result.get('dominant_gender', 'N/A')} ({max(result['gender'].values()):.2f}%)")
        print(f"Emotion: {result.get('dominant_emotion', 'N/A')}\n")
        print("Race Probabilities:")
        for race, score in result.get("race", {}).items():
            print(f"  {race:15}: {score:.2f}%")
        print(f"Most Likely Race: {result['dominant_race']}")
        print("-" * 40)

    annotated_img = draw_annotations(img, detections, results_list)

    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.title("Annotated Image with Age, Gender, Emotion")
    plt.show()


if __name__ == "__main__":
    main()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers

num_emotions = 7
num_ethnicities = 5
base_model = tf.keras.applications.VGG16(
    include_top=False,
    weights='imagenet',
    input_shape=(224, 224, 3)
)
base_model.trainable = False

x = base_model.output
x = layers.Flatten()(x)
x = layers.Dense(512, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.5)(x)

age_output = layers.Dense(1, name='age_output')(x)  # Regression (MSE/MAE)
gender_output = layers.Dense(1, activation='sigmoid', name='gender_output')(x)  # Binary
emotion_output = layers.Dense(num_emotions, activation='softmax', name='emotion_output')(x)  # Multi-class
ethnicity_output = layers.Dense(num_ethnicities, activation='softmax', name='ethnicity_output')(x)  # Multi-class

model_vgg = models.Model(
    inputs=base_model.input,
    outputs=[age_output, gender_output, emotion_output, ethnicity_output]
)

losses = {
    'age_output': 'mse',
    'gender_output': 'binary_crossentropy',
    'emotion_output': 'categorical_crossentropy',
    'ethnicity_output': 'categorical_crossentropy'
}

metrics = {
    'age_output': ['mae', 'mse'],
    'gender_output': ['accuracy'],
    'emotion_output': ['accuracy'],
    'ethnicity_output': ['accuracy']
}

# --- Compile with optimizer and regularization ---
model_vgg.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss=losses,
    metrics=metrics
)

model_vgg.summary()
```

```python
from google.colab import files
from itertools import combinations
from deepface import DeepFace
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Upload your test images (5 or fewer):")
uploaded = files.upload()

image_paths = list(uploaded.keys())
print(f"\nUploaded {len(image_paths)} images: {image_paths}")

if len(image_paths) < 2:
    raise Exception("Please upload at least 2 images.")

image_pairs = list(combinations(image_paths, 2))
print(f"\nGenerated {len(image_pairs)} pairs for testing.\n")

labels = []
print("For each pair, enter 1 if same person, 0 if different:")
for i, (img1, img2) in enumerate(image_pairs):
    print(f"Pair {i+1}: {img1} & {img2}")
    while True:
        label = input("Label (0/1): ")
        if label in ['0', '1']:
            labels.append(int(label))
            break
        else:
            print("Invalid input. Enter 0 or 1.")

y_true = np.array(labels)
y_pred = []
y_prob = []

print("\nRunning DeepFace verification...")
for img1, img2 in image_pairs:
    result = DeepFace.verify(img1_path=img1, img2_path=img2, enforce_detection=False)
    y_pred.append(int(result['verified']))
    y_prob.append(1 - result['distance'])

y_pred = np.array(y_pred)
y_prob = np.array(y_prob)

np.save("y_true.npy", y_true)
np.save("y_pred.npy", y_pred)
np.save("y_prob.npy", y_prob)
print("\nSaved y_true.npy, y_pred.npy, y_prob.npy")

cm = confusion_matrix(y_true, y_pred)
if cm.size == 4:
    tn, fp, fn, tp = cm.ravel()
else:
    tn = fp = fn = tp = 0
    if y_true[0] == y_pred[0] == 1: tp=1
    elif y_true[0] == y_pred[0] == 0: tn=1
    elif y_true[0]==1 and y_pred[0]==0: fn=1
    elif y_true[0]==0 and y_pred[0]==1: fp=1

accuracy = (tp + tn) / (tp + tn + fp + fn + 1e-8)
precision = tp / (tp + fp + 1e-8)
recall = tp / (tp + fn + 1e-8)
specificity = tn / (tn + fp + 1e-8)
f1 = 2 * precision * recall / (precision + recall + 1e-8)

plt.figure(figsize=(5,4))
sns.heatmap([[tn, fp],[fn, tp]], annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()

metrics = {
    "Accuracy": accuracy,
    "Precision": precision,
    "Specificity": specificity,
    "TP": tp,
    "FP": fp,
    "TN": tn,
    "FN": fn
}

from IPython.display import display
import pandas as pd
metrics_table = pd.DataFrame(list(metrics.items()), columns=["Metric", "Value"])
display(metrics_table)

plt.figure(figsize=(8,4))
main_metrics = ["Accuracy","Precision","Specificity"]
plt.bar(main_metrics, [metrics[m] for m in main_metrics], color='skyblue')
plt.ylim(0,1)
plt.title("Evaluation Metrics")
plt.ylabel("Value")
plt.show()
```
