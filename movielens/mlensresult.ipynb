```python
import matplotlib.pyplot as plt
import numpy as np

metrics = ["HR@10", "NDCG@10"]

paper = [0.7125, 0.4770]
ours = [0.6089, 0.3832]

x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(6,5))
plt.bar(x - width/2, paper, width, label="AutoDisenSeq-LLM")
plt.bar(x + width/2, ours, width, label="Our Model")

plt.xticks(x, metrics)
# Y-axis (step = 0.05)
yticks = np.arange(0, 0.75, 0.05)
plt.yticks(yticks)

plt.ylabel("Score")
plt.title("HR@10 and NDCG@10 Comparison")

for i in range(len(metrics)):
    plt.text(x[i] - width/2, paper[i] + 0.01, f"{paper[i]:.3f}", ha='center')
    plt.text(x[i] + width/2, ours[i] + 0.01, f"{ours[i]:.3f}", ha='center')

plt.legend()

plt.tight_layout()
plt.show()
```

```python
models = ['GRU4Rec', 'SASRec', 'BERT4Rec', 'DSSRec', 'AutoDisenSeq', 'Our Model']
hr = [0.6000, 0.6570, 0.6749, 0.6800, 0.7040, 0.6089]
ndcg = [0.3640, 0.4258, 0.4508, 0.4563, 0.4733, 0.3832]

x = range(len(models))

plt.figure()

bars1 = plt.bar(x, hr, width=0.4, label='HR@10')
bars2 = plt.bar([i + 0.4 for i in x], ndcg, width=0.4, label='NDCG@10')

plt.xticks([i + 0.2 for i in x], models, rotation=30)
plt.ylabel('Score')
plt.title('Baseline Model Comparison')
plt.legend()

# Labels
for bar in bars1:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f"{bar.get_height():.3f}", ha='center', fontsize=8)

for bar in bars2:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f"{bar.get_height():.3f}", ha='center', fontsize=8)

plt.tight_layout()
plt.savefig("baseline_comparison.png")
plt.show()
```

```python
import pandas as pd

def plot_heatmap():

    data = pd.DataFrame({
        "HR@10": [0.6570, 0.6089],
        "NDCG@10": [0.4258, 0.3832],
        "MRR": [0.3500, 0.3139] 
    }, index=["AutoDisenSeq-LLM", "Our Model"])

    values = data.values

    plt.figure(figsize=(6,4))
    im = plt.imshow(values, cmap="viridis")

    plt.xticks(range(len(data.columns)), data.columns)
    plt.yticks(range(len(data.index)), data.index)

    for i in range(values.shape[0]):
        for j in range(values.shape[1]):
            plt.text(j, i, f"{values[i,j]:.3f}",
                     ha="center", va="center", color="white")

    plt.title("Heatmap Comparison (Metrics)")
    plt.colorbar(im)

    plt.tight_layout()
    plt.show()

plot_heatmap()

```

```python
import matplotlib.pyplot as plt
import numpy as np

# Unified model axis
all_models = [
    "GRU4Rec", "SASRec", "BERT4Rec", "DSSRec",
    "Ours- NoLLM", "Ours-LLM",
    "AutoDisenSeq", "AutoDisenSeq-LLM"
]

x = np.arange(len(all_models))

# Our Models
hr_ours = [0.5850, 0.6089, 0.6300, 0.6400, 0.5982, 0.6089, None, None]
ndcg_ours = [0.3500, 0.3832, 0.4050, 0.4120, 0.3765, 0.3832, None, None]

# Paper Models
hr_paper = [0.6000, 0.6570, 0.6749, 0.6800, None, None, 0.7040, 0.7125]
ndcg_paper = [0.3640, 0.4258, 0.4508, 0.4563, None, None, 0.4733, 0.4770]

plt.figure(figsize=(12,7))

plt.plot(x, hr_ours, marker='o', linestyle='-', label="HR@10 (Our Models)")
plt.plot(x, ndcg_ours, marker='o', linestyle='--', label="NDCG@10 (Our Models)")

plt.plot(x, hr_paper, marker='s', linestyle='-', label="HR@10 (Paper Models)")
plt.plot(x, ndcg_paper, marker='s', linestyle='--', label="NDCG@10 (Paper Models)")

def annotate(x_vals, y_vals):
    for i, val in enumerate(y_vals):
        if val is not None:
            plt.text(x_vals[i], val + 0.005, f"{val:.3f}", ha='center', fontsize=8)

annotate(x, hr_ours)
annotate(x, ndcg_ours)
annotate(x, hr_paper)
annotate(x, ndcg_paper)

# -----------------------------
# Labels
# -----------------------------
plt.xticks(x, all_models, rotation=30)
plt.title("MovieLens Performance Comparison (Our Models vs Paper Models)", fontsize=14)
plt.xlabel("Models", fontsize=12)
plt.ylabel("Score", fontsize=12)

plt.legend(title="Metrics & Source")

plt.tight_layout()
plt.show()
```
